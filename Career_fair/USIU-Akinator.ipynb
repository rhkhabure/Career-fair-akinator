```python
import pandas as pd

# Simplified dataset for demo (just a few programs)
program_descriptions_business = {
    "International Business Administration": "Focuses on global trade, cross-border strategy, and leadership in international markets.",
    "Accounting": "Covers financial reporting, auditing, taxation, and professional CPA certification preparation.",
    "Finance": "Emphasizes investment analysis, valuation, and financial modeling for corporate and global markets.",
    "Hotel & Restaurant Management": "Centers on hospitality operations, service management, and guest experience design.",
    "Business Administration (MBA)": "Advanced management program focusing on leadership, strategy, and executive decision-making.",
    "Global Leadership & Management": "Explores cross-cultural leadership, organizational psychology, and global policy management.",
    "Global Banking and Finance": "Examines international financial systems, sovereign risk, and global market operations.",
    "Global Business Management": "Develops skills in multinational strategy, mergers, and global organizational management.",
    "Health Leadership and Management": "Integrates healthcare policy, hospital administration, and health economics leadership.",
    "Management and Organizational Development": "Focuses on organizational psychology, change management, and OD interventions.",
    "Doctor of Business Administration": "Research-intensive doctorate emphasizing applied executive research and advanced methodologies."
}
program_descriptions_humanities = {
    "International Relations": "Explores diplomacy, global politics, and policy analysis with practical simulations and regional studies.",
    "Psychology": "Introduces theories of human behavior, research methods, and experimental labs in cognitive and social psychology.",
    "Criminal Justice Studies": "Focuses on law enforcement, legal systems, and applied justice practices including internships and fieldwork.",
    "Criminal and Transitional Justice": "Examines justice in post-conflict societies, truth commissions, and restorative practices with policy focus.",
    "Sociology": "Studies social structures, community ethnography, and qualitative research methods in social theory.",
    "Clinical Psychology": "Prepares students for clinical practice with supervised therapy, evidence-based interventions, and licensure preparation.",
    "Counseling Psychology": "Emphasizes counseling techniques, career guidance, and practicum-based clinical supervision.",
    "International Relations (MA)": "Advanced study of diplomacy, global policy, and international relations research with thesis requirement.",
    "Marriage and Family Therapy": "Trains students in systemic therapy models, supervised practicum hours, and licensure track preparation.",
    "PhD International Relations": "Doctoral research in comparative politics, advanced theory, and archival methods for global policy analysis.",
    "Doctor of Psychology (Clinical)": "Doctorate focusing on advanced clinical research, longitudinal trials, and professional licensure in psychology."
}
program_descriptions_science_tech = {
    "Applied Computer Technology": "Focuses on hardware integration, embedded systems, and applied computing solutions with practical labs.",
    "Information Systems & Technology": "Covers enterprise systems, IT alignment with business processes, and systems integration projects.",
    "Data Science & Analytics": "Emphasizes big data platforms, statistical modeling, and feature engineering for predictive analysis.",
    "Artificial Intelligence & Robotics": "Explores robotics, control systems, and AI applications with hands-on sensor integration projects.",
    "Software Engineering": "Centers on software design, CI/CD practices, and building large-scale systems with engineering principles.",
    "Information Security": "Advanced study of cybersecurity, cryptography, and red-team labs with professional certification preparation.",
    "Information Systems and Technology": "Masters-level program focusing on enterprise architecture, IT governance, and systems security research."
}
program_descriptions_pharmacy = {
    "Bachelor of Pharmacy": "Prepares students for pharmaceutical practice with compounding, drug formulation, and licensure requirements.",
    "Nursing": "Focuses on patient care, clinical placements, and supervised practicum hours for professional licensure.",
    "Epidemiology & Biostatistics": "Covers public health surveillance, statistical analysis, and applied epidemiology projects.",
    "Analytical Chemistry": "Emphasizes laboratory techniques, chromatography, and analytical method validation in chemical sciences.",
    "Applied Biochemistry": "Explores protein assays, enzyme kinetics, and molecular biology techniques in applied laboratory settings.",
    "Clinical Pharmacology and Therapeutics": "Advanced study of drug regulation, clinical trials, and pharmacovigilance with licensure preparation."
}
program_descriptions_communication = {
    "Journalism": "Focuses on reporting, media writing, and newsroom practice with internships and investigative projects.",
    "Animation": "Covers visual effects, character design, and animation production workflows with studio-based projects.",
    "Film Production & Directing": "Explores cinematography, directing, and post-production pipelines with hands-on studio internships.",
    "Communication Studies": "Advanced program in media research, editorial strategy, and communication theory with thesis requirement."
}

# Here we are doing embeddings for the programs to help the model understand their content and differentiate them based on their descriptions, attributes, and other metadata. This will allow the model to make more informed decisions when matching candidates to programs based on their responses to the questions.
for program in programs:
    program["embedding_text"] = (
        f"{program['name']} ({program['level']}, {program['school']}) "
        f"Attributes: {', '.join(program['attributes'])}. "
        f"Description: {program['description']}"
    )
    


# ----------------------------
# Question bank for Akinator
# ----------------------------
import math

def entropy(counts):
    total = sum(counts.values())
    if total == 0: return 0
    return -sum((c/total) * math.log2(c/total) for c in counts.values() if c > 0)

def information_gain(candidates, question, programs):
    # candidates: list of program dicts still possible
    # question: question dict
    # programs: full program list
    
    # For each possible response, count how many candidates match
    response_counts = {}
    for resp in question["responses"]:
        response_counts[resp] = 0
        for prog in candidates:
            # if program has any of the question tags, assume it could map to this response
            if any(tag in prog["attributes"] for tag in question["tags"]):
                response_counts[resp] += 1
    
    # Compute entropy before asking
    base_entropy = entropy({prog["name"]:1 for prog in candidates})
    # Compute entropy after asking
    weighted_entropy = sum((sum(response_counts.values())/len(candidates)) * entropy(response_counts) for resp in response_counts)
    
    return base_entropy - weighted_entropy

def select_next_question(candidates, asked_questions, questions, programs):
    unused = {qid:meta for qid,meta in questions.items() if qid not in asked_questions}
    gains = {qid: information_gain(candidates, meta, programs) for qid,meta in unused.items()}
    return max(gains, key=gains.get)

path_log = {
    "path": [("q_coding","Definitely Yes"),("q_lab_work","Probably No")],
    "final_guess": "Applied Computer Technology",
    "correct_program": "Data Science & Analytics"
}
```

```python
import pandas as pd

# Load full dataset
df = pd.DataFrame(programs)  # programs = full expanded list
df["score"] = 0.0

response_weights = {
    "Definitely Yes": 1.0,
    "Probably Yes": 0.75,
    "Neutral": 0.5,
    "Probably No": 0.25,
    "Definitely No": 0.1
}

def update_scores(candidates, attr, response):
    weight = response_weights[response]
    for i, row in candidates.iterrows():
        if attr in row["attributes"]:
            candidates.at[i, "score"] += weight
        else:
            candidates.at[i, "score"] += (1 - weight) * 0.2
    return candidates

def pick_next_question(candidates, asked):
    # Collect all attributes from remaining candidates
    all_attrs = set(sum(candidates["attributes"], []))
    # Remove already asked
    available_attrs = [a for a in all_attrs if a not in asked]
    
    # Find attribute with best split (closest to 50/50)
    best_attr, best_balance = None, 1.0
    for attr in available_attrs:
        with_attr = candidates[candidates["attributes"].apply(lambda x: attr in x)]
        without_attr = candidates[candidates["attributes"].apply(lambda x: attr not in x)]
        balance = abs(len(with_attr) - len(without_attr)) / len(candidates)
        if balance < best_balance:
            best_balance = balance
            best_attr = attr
    return best_attr


```

```python
# Config (small for quick test; scale up later)
BASE_SIMS_PER_PROGRAM = 20  # increase for final dataset
SIMS_PER_PROGRAM_FOR_VALIDATION = 75
NOISE_RATE = 0.12
MAX_QUESTIONS = 10
LIKERT = ["Definitely Yes","Probably Yes","Neutral","Probably No","Definitely No"]
RESP_ENC = {"Definitely Yes": 1.0, "Probably Yes": 0.75, "Neutral": 0.5, "Probably No": 0.25, "Definitely No": 0.0}
```

```python
# Build attribute vocabulary
all_attrs = sorted({a for p in programs for a in p["attributes"]})

def path_to_features(path, all_attrs):
    attr_scores = {a: 0.0 for a in all_attrs}
    for attr, response in path:
        val = RESP_ENC.get(response, 0.5)
        attr_scores[attr] = max(attr_scores.get(attr, 0.0), val)
    num_q = len(path)
    avg_resp = np.mean([RESP_ENC.get(r, 0.5) for _, r in path]) if path else 0.0
    vals = [RESP_ENC.get(r, 0.5) for _, r in path]
    if len(vals) > 1:
        p = np.array(vals) / (np.sum(vals) + 1e-9)
        entropy = -np.sum(p * np.log(p + 1e-9))
    else:
        entropy = 0.0
    feat = [attr_scores[a] for a in all_attrs] + [num_q, avg_resp, entropy]
    return np.array(feat, dtype=float)

```

```python

# Generate synthetic dataset (balanced by school)
rows = []
per_school_multiplier = compute_default_school_multiplier(programs_df)
for prog in programs:
    school = prog["school"]
    multiplier = per_school_multiplier.get(school, 1.0)
    sims = max(1, int(BASE_SIMS_PER_PROGRAM * multiplier))
    for _ in range(sims):
        qids, path, candidates = simulate_path_for_program(prog, MAX_QUESTIONS, {}, noise_rate=NOISE_RATE)
        feat = path_to_features(path, all_attrs)
        rows.append({"program": prog["name"], "school": school, "path": path, "features": feat})

# Build arrays
X = np.vstack([r["features"] for r in rows])
y = np.array([r["program"] for r in rows])
meta = pd.DataFrame([{"program": r["program"], "school": r["school"], "path": r["path"]} for r in rows])

# Train/test split (stratify by program)
X_train, X_test, y_train, y_test, meta_train, meta_test = train_test_split(
    X, y, meta, test_size=0.2, random_state=42, stratify=y
)
```

```python

# Train baseline classifier
clf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
clf.fit(X_train, y_train)

# Evaluation helpers
def top_k_accuracy(clf, X, y_true, k=3):
    probs = clf.predict_proba(X)
    classes = clf.classes_
    topk = np.argsort(probs, axis=1)[:, ::-1][:, :k]
    topk_names = classes[topk]
    hits = [1 if y_true[i] in topk_names[i] else 0 for i in range(len(y_true))]
    return np.mean(hits)

# Evaluate
y_pred = clf.predict(X_test)
acc = accuracy_score(y_test, y_pred)
top3 = top_k_accuracy(clf, X_test, y_test, k=3)
print(f"Test accuracy (top-1): {acc:.4f}")
print(f"Test top-3 accuracy: {top3:.4f}")

# Correct per-school breakdown
per_school = {}
for school, group_df in meta_test.groupby("school"):
    # group_df is the subset of meta_test rows for this school
    # get positional indices relative to meta_test
    idxs = group_df.index.get_indexer(group_df.index)  # or simply use group_df.index.to_numpy()
    
    # but simpler: just slice y_test and X_test with group_df.index positions
    y_true_school = y_test[group_df.index.get_indexer(meta_test.index)]
    X_school = X_test[group_df.index.get_indexer(meta_test.index)]
    
    y_pred_school = clf.predict(X_school)
    acc_school = accuracy_score(y_true_school, y_pred_school)
    top3_school = top_k_accuracy(clf, X_school, y_true_school, k=3)
    per_school[school] = {"top1": acc_school, "top3": top3_school, "n": len(group_df)}

pd.DataFrame.from_dict(per_school, orient="index").sort_values("top1", ascending=False)

# Top misclassification pairs
pairs = Counter()
for t, p in zip(y_test, y_pred):
    if t != p:
        pairs[(t,p)] += 1
print("\nTop misclassification pairs:")
for (t,p), cnt in pairs.most_common(20):
    print(f"{t} -> {p} : {cnt}")

# Save model and artifacts
os.makedirs("pretrained", exist_ok=True)
joblib.dump(clf, "pretrained/akinator_rf.joblib")
np.save("pretrained/X.npy", X)
np.save("pretrained/y.npy", y)
meta.to_pickle("pretrained/meta.pkl")
print("\nSaved model and dataset to ./pretrained/")
```
